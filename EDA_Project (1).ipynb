# Data Cleaning

```python
import pandas as pd
import numpy as np

def clean_currency_to_int(value):
    # 1. Handle missing values first so the code doesn't crash
    if pd.isna(value):
        return np.nan

    # 2. Ensure it's a string, then remove the '$' and any extra spaces
    val_str = str(value).replace('$', '').strip()

    # 3. Check for suffixes, multiply, and convert to integer
    try:
        if val_str.endswith('B'):
            # float() is used first in case of decimals like "$1.5B"
            return int(float(val_str[:-1]) * 1_000_000_000)

        elif val_str.endswith('M'):
            return int(float(val_str[:-1]) * 1_000_000)

        elif val_str.endswith('K'):
            return int(float(val_str[:-1]) * 1_000)

        else:
            # If there's no suffix, just convert the raw number
            return int(float(val_str))

    except ValueError:
        # If the string contains junk data (e.g., "Not Available"), return NaN
        return np.nan

# Apply the cleaning function to 'Valuation' and 'Funding' columns
df['Valuation_Cleaned'] = df['Valuation'].apply(clean_currency_to_int)
df['Funding_Cleaned'] = df['Funding'].apply(clean_currency_to_int)
```

# Data Plot Visualisation

# Description of each graph
### Data Visualization Methods

* **Bar Plot (Top Investors)**
  Uses Seaborn (`sns.barplot`) to visualize the top 15 most active investors in the dataset. It counts the number of unicorn companies each investor has backed, highlighting the key players driving startup funding.

* **Tree Map (Valuation Hierarchy)**
  Uses Plotly (`px.treemap`) to display the hierarchical distribution of unicorn valuations. The map starts from the global level ("World"), breaks down into continents, and further categorizes by industry. The size of each rectangle represents the total valuation of companies in that sector.

* **Sankey Diagram (Capital Flow)**
  Uses Plotly (`go.Sankey`) to illustrate the directional flow of capital. It connects continents (source) to specific industries (target), visually demonstrating how investment values are distributed across different global regions and sectors.

* **Bar Chart (Top Countries by Unicorn Count)**
  A standard bar chart visualizing the countries with the highest concentration of unicorn startups, showcasing the dominance of regions like the US, China, and India.

* **Bar Chart (Top Industries)**
  Displays the frequency of unicorns across different sectors (e.g., Fintech, E-commerce, Artificial Intelligence), identifying which markets are currently the most lucrative or saturated.

* **Line Chart (Unicorns Joined per Year)**
  A time-series plot tracking the number of companies that reached unicorn status each year. This highlights periods of rapid market growth or investment booms.

* **Histogram (Years to Unicorn Status)**
  Shows the distribution of time (in years) it took for companies to reach a $1B valuation from their founding date, providing insight into the "speed to success" for the average startup.

* **Scatter Plot (Funding vs. Valuation)**
  Plots total capital raised against current market valuation. This helps identify if more funding strictly correlates to a higher valuation, or if certain companies are highly valued despite lower funding.

* **Box Plot (Valuation Distribution by Industry)**
  Highlights the spread, median, and outliers of valuations within individual industries, easily pinpointing "Elite Outliers" (like Bytedance or SpaceX) that heavily skew their respective sectors.

* **Pie / Donut Chart (Geographic Distribution)**
  Illustrates the proportion of total unicorns located in each continent, giving a quick, high-level overview of the global startup ecosystem.

* **Correlation Heatmap**
  Uses Seaborn (`sns.heatmap`) to visualize the correlation matrix between numerical variables (such as Year Founded, Valuation, Funding, and Years to Unicorn), revealing hidden statistical relationships in the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from scipy.cluster.hierarchy import linkage, dendrogram
from matplotlib_venn import venn2
import plotly.graph_objects as go
import networkx as nx
```

```python
# 1. Load the dataset
df = pd.read_csv('Unicorn_Companies.csv')

# 2. Clean the 'Valuation' column (convert to Billions as float)
df['Valuation_B'] = df['Valuation'].str.replace('$', '', regex=False).str.replace('B', '', regex=False).astype(float)

# 3. Clean the 'Funding' column (handle Billions 'B', Millions 'M', and 'Unknown' values)
def clean_funding(x):
    if pd.isna(x) or x == 'Unknown':
        return np.nan
    x = str(x).replace('$', '')
    if 'B' in x: return float(x.replace('B', '')) * 1e9
    if 'M' in x: return float(x.replace('M', '')) * 1e6
    if 'K' in x: return float(x.replace('K', '')) * 1e3
    return float(x)

df['Funding_Cleaned'] = df['Funding'].apply(clean_funding)

# 4. Convert 'Date Joined' to datetime and calculate 'Years to Unicorn'
df['Date Joined'] = pd.to_datetime(df['Date Joined'])
df['Year Joined'] = df['Date Joined'].dt.year
df['Years to Unicorn'] = df['Year Joined'] - df['Year Founded']

# Set a unified style for seaborn plots
sns.set_theme(style="whitegrid")
```

```python
# Outliers

# 2. Data Cleaning: Convert 'Valuation' and 'Funding' from strings to numeric values
def clean_currency(val):
    if pd.isna(val) or val == 'Unknown':
        return np.nan
    val = str(val).replace('$', '')
    if 'B' in val:
        return float(val.replace('B', '')) * 1e9  # Convert Billions
    elif 'M' in val:
        return float(val.replace('M', '')) * 1e6  # Convert Millions
    elif 'K' in val:
        return float(val.replace('K', '')) * 1e3  # Convert Thousands
    else:
        return float(val)

df['Valuation_Cleaned'] = df['Valuation'].apply(clean_currency)
df['Funding_Cleaned'] = df['Funding'].apply(clean_currency)

# 3. Define a function to get outliers using the IQR method
def get_outliers(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter the dataframe for outliers
    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    return outliers

# 4. Extract outliers for each numerical column
val_outliers = get_outliers(df, 'Valuation_Cleaned')
fund_outliers = get_outliers(df, 'Funding_Cleaned')
year_outliers = get_outliers(df, 'Year Founded')

print(f"Total Valuation Outliers: {len(val_outliers)}")
print(f"Total Funding Outliers: {len(fund_outliers)}")
print(f"Total Year Founded Outliers: {len(year_outliers)}")

# ==========================================
# 5. Visualization for EDA (Boxplots)
# ==========================================
plt.figure(figsize=(18, 5))

# Boxplot for Valuation
plt.subplot(1, 3, 1)
sns.boxplot(x=df['Valuation_Cleaned'])
plt.title('Boxplot of Valuation (Detecting Outliers)')
plt.xlabel('Valuation (in Billions)')

# Boxplot for Funding
plt.subplot(1, 3, 2)
sns.boxplot(x=df['Funding_Cleaned'])
plt.title('Boxplot of Funding (Detecting Outliers)')
plt.xlabel('Funding (in Billions)')

# Boxplot for Year Founded
plt.subplot(1, 3, 3)
sns.boxplot(x=df['Year Founded'])
plt.title('Boxplot of Year Founded (Detecting Outliers)')
plt.xlabel('Year Founded')

plt.tight_layout()
plt.show()

# Optional: View the top 5 valuation outliers
# display(val_outliers[['Company', 'Valuation_Cleaned']].sort_values(by='Valuation_Cleaned', ascending=False).head())
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

# --- 1. Load and Clean Data ---
df = pd.read_csv('Unicorn_Companies.csv')

# Function to parse string valuations into numbers
def parse_money(m):
    if pd.isna(m) or m == 'Unknown':
        return np.nan
    m = str(m).replace('$', '')
    if 'B' in m: return float(m.replace('B', '')) * 1e9
    if 'M' in m: return float(m.replace('M', '')) * 1e6
    return float(m)

df['Valuation_num'] = df['Valuation'].apply(parse_money)

# --- 2. Prepare Data for Sankey ---
# To keep the diagram readable, filter for the top 6 Continents and top 5 Industries
top_conts = df['Continent'].value_counts().nlargest(6).index # Changed to nlargest(6) to include all continents
top_inds = df['Industry'].value_counts().nlargest(5).index

# Filter dataset to only include these top categories
sankey_data = df[df['Continent'].isin(top_conts) & df['Industry'].isin(top_inds)]

# Group the data to get the total valuation flowing from each Continent -> Industry
sankey_grouped = sankey_data.groupby(['Continent', 'Industry'])['Valuation_num'].sum().reset_index()

# Extract unique node names (Continents on the left, Industries on the right)
all_nodes = list(sankey_grouped['Continent'].unique()) + list(sankey_grouped['Industry'].unique())

# Create a mapping dictionary { 'Node_Name': Index_Number }
# Plotly requires links to refer to node indices (0, 1, 2, etc.) rather than names
node_indices = {node: i for i, node in enumerate(all_nodes)}

# Map the Continents and Industries to their respective index lists
source = sankey_grouped['Continent'].map(node_indices).tolist()
target = sankey_grouped['Industry'].map(node_indices).tolist()
value = sankey_grouped['Valuation_num'].tolist()

# --- 3. Build and Display the Sankey Diagram ---
fig_sankey = go.Figure(data=[go.Sankey(
    # Configuration for the rectangular nodes
    node = dict(
        pad=15,
        thickness=20,
        line=dict(color="white", width=0.5),
        label=all_nodes,
        color="lavender"
    ),
    # Configuration for the flowing links between nodes
    link = dict(
        source=source,
        target=target,
        value=value,
        color="rgba(65, 105, 225, 0.4)" # Royal blue with 40% opacity
    )
)])

# Update layout to add a title
fig_sankey.update_layout(
    title_text="Sankey Diagram: Total Valuation Flow from Continents to Industries",
    font_size=12,
    width=900,
    height=600
)

# Open the interactive plot in your default web browser
fig_sankey.show()

# Optional: You can also save it as an interactive HTML file
# fig_sankey.write_html('sankey_diagram.html')
```
